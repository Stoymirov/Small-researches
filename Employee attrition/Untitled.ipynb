```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
sns.countplot(data=df,x="Attrition")
plt.title("Employee Attrition Count")
plt.xlabel("Attrition")
plt.ylabel("Count")
plt.show()
```

```python
sns.countplot(data=df,x="JobRole",hue="Attrition")
plt.xticks(rotation=90)
plt.title("Attrition by role")
plt.tight_layout()
plt.show()
```

From the data above we can observe that the laboratory Technicians are most likely to leave, whereas **Research Director, Manager** are least likely

```python
sns.boxplot(data=df,x="Attrition",y="DistanceFromHome")
plt.title("Attrition by distance from home")
plt.show()
```

```python
sns.boxplot(data=df,x="Attrition",y="MonthlyIncome")
plt.title("Monthly Income by attrition")
plt.show()
```

## Gender analysis

From the boxplot it is visible that the salary contributes highly to attrition.
